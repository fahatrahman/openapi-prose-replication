````python
import hashlib
import os
import tempfile
from datetime import datetime, timezone

import yaml


def atomic_write_bytes(path: Path, data: bytes) -> None:
    path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    descriptor, temporary_name = tempfile.mkstemp(
        dir=path.parent,
        prefix=f".{path.name}.",
        suffix=".tmp",
    )

    try:
        with os.fdopen(descriptor, "wb") as handle:
            handle.write(data)
            handle.flush()
            os.fsync(handle.fileno())

        os.replace(temporary_name, path)

    except Exception:
        try:
            os.unlink(temporary_name)
        except FileNotFoundError:
            pass
        raise


def write_text(path: Path, text: str) -> None:
    atomic_write_bytes(
        path,
        text.encode("utf-8"),
    )


def write_json(path: Path, value) -> None:
    serialized = json.dumps(
        value,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    ) + "\n"

    write_text(path, serialized)


def write_yaml(path: Path, value) -> None:
    serialized = yaml.safe_dump(
        value,
        sort_keys=False,
        allow_unicode=True,
    )

    write_text(path, serialized)


def sha256_bytes(data: bytes) -> str:
    return hashlib.sha256(data).hexdigest()


def sha256_file(
    path: Path,
    chunk_size: int = 1024 * 1024,
) -> str:
    digest = hashlib.sha256()

    with path.open("rb") as handle:
        while chunk := handle.read(chunk_size):
            digest.update(chunk)

    return digest.hexdigest()


io_evidence_path = (
    PROJECT_ROOT
    / "reports"
    / "runtime"
    / "io_qualification.json"
)

io_evidence = {
    "captured_utc": datetime.now(timezone.utc).isoformat(),
    "encoding": "UTF-8",
    "atomic_replacement": True,
    "hash_algorithm": "SHA-256",
    "status": "PASS",
}

write_json(io_evidence_path, io_evidence)

loaded_io_evidence = json.loads(
    io_evidence_path.read_text(encoding="utf-8")
)

io_report = {
    "path": str(io_evidence_path),
    "file_exists": io_evidence_path.is_file(),
    "round_trip_equal": loaded_io_evidence == io_evidence,
    "sha256": sha256_file(io_evidence_path),
}

print(json.dumps(io_report, indent=2))

if not io_report["file_exists"]:
    raise RuntimeError("Atomic file-writing test failed.")

if not io_report["round_trip_equal"]:
    raise RuntimeError("JSON round-trip verification failed.")

print("\nCELL 5: PASS")
````

````python
def canonical_json_text(document) -> str:
    return json.dumps(
        document,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
    )


def prompt_boundary(
    study_id: str,
    operation_id: str,
) -> str:
    payload = (
        f"{study_id}|{operation_id}"
        .encode("utf-8")
    )

    return hashlib.sha256(
        payload
    ).hexdigest()[:24]


def normalize_token_ids(token_result) -> list[int]:
    # Transformers 5 may return a BatchEncoding.
    if (
        hasattr(token_result, "keys")
        and "input_ids" in token_result
    ):
        token_result = token_result[
            "input_ids"
        ]

    # Handle tensors and NumPy-like arrays.
    if hasattr(token_result, "tolist"):
        token_result = token_result.tolist()

    # Remove a single batch dimension.
    if (
        token_result
        and isinstance(token_result[0], list)
    ):
        if len(token_result) != 1:
            raise RuntimeError(
                "Expected exactly one rendered prompt."
            )

        token_result = token_result[0]

    if not isinstance(token_result, list):
        raise TypeError(
            "Unsupported token result type: "
            f"{type(token_result).__name__}"
        )

    return [
        int(token_id)
        for token_id in token_result
    ]


def render_development_prompt():
    operation_id = (
        "development-fixture::post::/users"
    )

    boundary = prompt_boundary(
        STUDY_ID,
        operation_id,
    )

    canonical_json = canonical_json_text(
        development_fixture
    )

    if boundary in canonical_json:
        raise RuntimeError(
            "The prompt boundary appears inside "
            "the canonical OpenAPI JSON."
        )

    user_prompt = (
        USER_PROMPT_TEMPLATE.format(
            BOUNDARY=boundary,
            CANONICAL_OPENAPI_JSON=(
                canonical_json
            ),
        )
    )

    messages = [
        {
            "role": "system",
            "content": SYSTEM_PROMPT,
        },
        {
            "role": "user",
            "content": user_prompt,
        },
    ]

    rendered_text = (
        tokenizer.apply_chat_template(
            messages,
            tokenize=False,
            add_generation_prompt=True,
        )
    )

    token_result = (
        tokenizer.apply_chat_template(
            messages,
            tokenize=True,
            add_generation_prompt=True,
        )
    )

    token_ids = normalize_token_ids(
        token_result
    )

    return {
        "operation_id": operation_id,
        "boundary": boundary,
        "canonical_json": canonical_json,
        "user_prompt": user_prompt,
        "rendered_text": rendered_text,
        "rendered_bytes": (
            rendered_text.encode("utf-8")
        ),
        "token_ids": token_ids,
    }


development_fixture = {
    "openapi": "3.0.3",

    "info": {
        "title": "Development fixture",
        "version": "1",
    },

    "paths": {
        "/users": {
            "post": {
                "summary": "Create a user",

                "requestBody": {
                    "required": True,
                    "content": {
                        "application/json": {
                            "schema": {
                                "type": "object",
                                "required": ["age"],
                                "properties": {
                                    "age": {
                                        "type": "integer",
                                        "minimum": 18,
                                        "maximum": 60,
                                    },
                                },
                            },
                        },
                    },
                },

                "responses": {
                    "201": {
                        "description": "User created",
                        "content": {
                            "application/json": {
                                "schema": {
                                    "type": "object",
                                    "required": ["id"],
                                    "properties": {
                                        "id": {
                                            "type": "string",
                                        },
                                    },
                                },
                            },
                        },
                    },

                    "400": {
                        "description": (
                            "Invalid age input"
                        ),
                        "content": {
                            "application/json": {
                                "schema": {
                                    "type": "object",
                                    "required": ["error"],
                                    "properties": {
                                        "error": {
                                            "type": "string",
                                        },
                                    },
                                },
                            },
                        },
                    },
                },
            },
        },
    },
}


development_fixture_root = (
    PROJECT_ROOT
    / "data"
    / "development_examples"
)

development_fixture_root.mkdir(
    parents=True,
    exist_ok=True,
)

development_fixture_path = (
    development_fixture_root
    / "users.json"
)

write_text(
    development_fixture_path,
    canonical_json_text(
        development_fixture
    ) + "\n",
)


first_render = render_development_prompt()
second_render = render_development_prompt()

rendered_prompt_path = (
    PROJECT_ROOT
    / "reports"
    / "runtime"
    / "development_prompt.txt"
)

rendered_token_ids_path = (
    PROJECT_ROOT
    / "reports"
    / "runtime"
    / "development_prompt_token_ids.json"
)

prompt_evidence_path = (
    PROJECT_ROOT
    / "reports"
    / "runtime"
    / "prompt_reproducibility.json"
)

write_text(
    rendered_prompt_path,
    first_render["rendered_text"],
)

write_json(
    rendered_token_ids_path,
    first_render["token_ids"],
)


prompt_checks = {
    "rendered_bytes_identical": (
        first_render["rendered_bytes"]
        == second_render["rendered_bytes"]
    ),

    "token_ids_identical": (
        first_render["token_ids"]
        == second_render["token_ids"]
    ),

    "boundary_absent_from_json": (
        first_render["boundary"]
        not in first_render["canonical_json"]
    ),

    "development_prompt_within_limit": (
        len(first_render["token_ids"])
        <= MAX_RENDERED_INPUT_TOKENS
    ),
}

prompt_evidence = {
    "status": (
        "PASS"
        if all(prompt_checks.values())
        else "BLOCKED"
    ),

    "checks": prompt_checks,

    "operation_id": (
        first_render["operation_id"]
    ),
    "boundary": first_render["boundary"],

    "rendered_prompt_tokens": len(
        first_render["token_ids"]
    ),

    "configured_input_limit": (
        MAX_RENDERED_INPUT_TOKENS
    ),

    "rendered_prompt_sha256": (
        sha256_bytes(
            first_render["rendered_bytes"]
        )
    ),

    "token_ids_sha256": sha256_bytes(
        json.dumps(
            first_render["token_ids"],
            separators=(",", ":"),
        ).encode("utf-8")
    ),

    "development_fixture_sha256": (
        sha256_file(
            development_fixture_path
        )
    ),
}

write_json(
    prompt_evidence_path,
    prompt_evidence,
)

print(json.dumps(prompt_evidence, indent=2))

if prompt_evidence["status"] != "PASS":
    raise RuntimeError(
        "Prompt reproducibility checks failed."
    )

print("\nCELL 12: PASS")
````

````python
import re


OUTER_PYTHON_FENCE = re.compile(
    (
        r"\A\s*```python[ \t]*\r?\n"
        r"(?P<code>[\s\S]*?)"
        r"\r?\n```\s*\Z"
    ),
    flags=re.IGNORECASE,
)


def extract_python_module(
    raw_text: str,
):
    fence_match = (
        OUTER_PYTHON_FENCE.fullmatch(
            raw_text
        )
    )

    if fence_match:
        extracted_code = (
            fence_match.group("code")
        )

        if "```" in extracted_code:
            return (
                None,
                "rejected",
                "nested_or_multiple_fences",
            )

        return (
            extracted_code,
            "accepted_outer_python_fence",
            None,
        )

    if "```" in raw_text:
        return (
            None,
            "rejected",
            "mixed_or_multiple_fences",
        )

    if not raw_text.strip():
        return (
            None,
            "rejected",
            "empty_output",
        )

    return (
        raw_text,
        "accepted_raw_python",
        None,
    )


extraction_policy = {
    "policy_version": "1.0",

    "accepted": [
        (
            "Raw Python text containing no "
            "Markdown fence."
        ),
        (
            "One complete outer ```python "
            "fence with no other fences."
        ),
    ],

    "rejected": [
        "Empty output.",
        "Multiple code blocks.",
        (
            "Explanatory text mixed with "
            "a fenced module."
        ),
        (
            "Non-Python or incomplete "
            "Markdown fences."
        ),
    ],

    "repairs_permitted": False,
    "raw_output_preserved": True,

    "notes": [
        (
            "Extraction does not establish "
            "that the source compiles."
        ),
        (
            "Compilation is measured later "
            "as a separate diagnostic."
        ),
    ],
}


extraction_test_cases = [
    {
        "case_id": "raw_python",
        "raw": (
            "import pytest\n\n"
            "def test_ok():\n"
            "    assert True\n"
        ),
        "expected_status": (
            "accepted_raw_python"
        ),
    },

    {
        "case_id": "one_outer_fence",
        "raw": (
            "```python\n"
            "import pytest\n\n"
            "def test_ok():\n"
            "    assert True\n"
            "```"
        ),
        "expected_status": (
            "accepted_outer_python_fence"
        ),
    },

    {
        "case_id": "mixed_explanation",
        "raw": (
            "Here is the module:\n"
            "```python\n"
            "assert True\n"
            "```"
        ),
        "expected_status": "rejected",
    },

    {
        "case_id": "multiple_blocks",
        "raw": (
            "```python\n"
            "assert True\n"
            "```\n"
            "```python\n"
            "assert False\n"
            "```"
        ),
        "expected_status": "rejected",
    },

    {
        "case_id": "non_python_fence",
        "raw": (
            "```text\n"
            "assert True\n"
            "```"
        ),
        "expected_status": "rejected",
    },

    {
        "case_id": "empty_output",
        "raw": "   \n",
        "expected_status": "rejected",
    },
]


extraction_results = []

for test_case in extraction_test_cases:
    extracted, status, reason = (
        extract_python_module(
            test_case["raw"]
        )
    )

    extraction_results.append({
        "case_id": test_case["case_id"],
        "expected_status": (
            test_case["expected_status"]
        ),
        "observed_status": status,
        "reason": reason,
        "pass": (
            status
            == test_case["expected_status"]
        ),
        "extracted_sha256": (
            sha256_bytes(
                extracted.encode("utf-8")
            )
            if extracted is not None
            else None
        ),
    })


extraction_qualification = {
    "status": (
        "PASS"
        if all(
            result["pass"]
            for result in extraction_results
        )
        else "BLOCKED"
    ),
    "policy_version": "1.0",
    "test_count": len(
        extraction_results
    ),
    "results": extraction_results,
}


extraction_policy_path = (
    PROJECT_ROOT
    / "reports"
    / "runtime"
    / "extraction_policy.json"
)

extraction_qualification_path = (
    PROJECT_ROOT
    / "reports"
    / "runtime"
    / "extraction_qualification.json"
)

write_json(
    extraction_policy_path,
    extraction_policy,
)

write_json(
    extraction_qualification_path,
    extraction_qualification,
)


print(
    json.dumps(
        extraction_qualification,
        indent=2,
    )
)

if (
    extraction_qualification["status"]
    != "PASS"
):
    raise RuntimeError(
        "Extraction-policy tests failed."
    )

print("\nCELL 13: PASS")
````

````python
import copy
import gc
import sys


# Release memory retained after the previous OOM.
globals().pop("long_probe_output", None)

sys.last_traceback = None

gc.collect()

for device_index in range(
    torch.cuda.device_count()
):
    torch.cuda.empty_cache()

torch.cuda.synchronize()


PREFILL_CHUNK_SIZE = 512


def render_fixture_document(
    document,
    operation_id: str,
):
    boundary = prompt_boundary(
        STUDY_ID,
        operation_id,
    )

    canonical_json = canonical_json_text(
        document
    )

    if boundary in canonical_json:
        raise RuntimeError(
            "Prompt boundary occurred in "
            "the synthetic document."
        )

    user_prompt = (
        USER_PROMPT_TEMPLATE.format(
            BOUNDARY=boundary,
            CANONICAL_OPENAPI_JSON=(
                canonical_json
            ),
        )
    )

    messages = [
        {
            "role": "system",
            "content": SYSTEM_PROMPT,
        },
        {
            "role": "user",
            "content": user_prompt,
        },
    ]

    rendered_text = (
        tokenizer.apply_chat_template(
            messages,
            tokenize=False,
            add_generation_prompt=True,
        )
    )

    token_result = (
        tokenizer.apply_chat_template(
            messages,
            tokenize=True,
            add_generation_prompt=True,
        )
    )

    return {
        "boundary": boundary,
        "canonical_json": canonical_json,
        "rendered_text": rendered_text,
        "token_ids": normalize_token_ids(
            token_result
        ),
    }


# Construct the largest synthetic prompt that
# does not exceed the frozen 5,120-token limit.
lower_repetitions = 0
upper_repetitions = 10000

best_long_render = None
best_repetition_count = None

while (
    lower_repetitions
    <= upper_repetitions
):
    midpoint = (
        lower_repetitions
        + upper_repetitions
    ) // 2

    candidate_fixture = copy.deepcopy(
        development_fixture
    )

    candidate_fixture[
        "paths"
    ][
        "/users"
    ][
        "post"
    ][
        "description"
    ] = (
        "qualification "
        * midpoint
    ).strip()

    candidate_render = (
        render_fixture_document(
            candidate_fixture,
            (
                "development-fixture::"
                "long-context::post::/users"
            ),
        )
    )

    candidate_token_count = len(
        candidate_render["token_ids"]
    )

    if (
        candidate_token_count
        <= MAX_RENDERED_INPUT_TOKENS
    ):
        best_long_render = (
            candidate_render
        )

        best_repetition_count = midpoint
        lower_repetitions = midpoint + 1

    else:
        upper_repetitions = midpoint - 1


if best_long_render is None:
    raise RuntimeError(
        "Could not construct a synthetic "
        "prompt within the input limit."
    )


long_input_token_count = len(
    best_long_render["token_ids"]
)

long_input_ids = torch.tensor(
    [best_long_render["token_ids"]],
    dtype=torch.long,
    device=input_device,
)

long_attention_mask = torch.ones_like(
    long_input_ids
)


memory_before_long_probe = []

for device_index in range(
    torch.cuda.device_count()
):
    torch.cuda.reset_peak_memory_stats(
        device_index
    )

    properties = (
        torch.cuda.get_device_properties(
            device_index
        )
    )

    allocated_bytes = (
        torch.cuda.memory_allocated(
            device_index
        )
    )

    memory_before_long_probe.append({
        "index": device_index,

        "total_bytes": int(
            properties.total_memory
        ),

        "allocated_bytes": int(
            allocated_bytes
        ),

        "free_estimate_bytes": int(
            properties.total_memory
            - allocated_bytes
        ),
    })


torch.cuda.synchronize()

long_probe_started = (
    time.perf_counter()
)

with torch.inference_mode():
    long_probe_output = model.generate(
        input_ids=long_input_ids,

        attention_mask=(
            long_attention_mask
        ),

        max_new_tokens=MAX_NEW_TOKENS,
        min_new_tokens=MAX_NEW_TOKENS,
        do_sample=False,

        pad_token_id=(
            tokenizer.pad_token_id
        ),

        eos_token_id=(
            tokenizer.eos_token_id
        ),

        use_cache=True,

        # Transformers 5 chunked prefill.
        prefill_chunk_size=(
            PREFILL_CHUNK_SIZE
        ),

        cache_implementation="dynamic",
    )

torch.cuda.synchronize()

long_probe_duration_seconds = (
    time.perf_counter()
    - long_probe_started
)


long_probe_generated_tokens = int(
    long_probe_output.shape[1]
    - long_input_ids.shape[1]
)


memory_after_long_probe = [
    {
        "index": device_index,

        "allocated_gib": round(
            torch.cuda.memory_allocated(
                device_index
            ) / (1024 ** 3),
            3,
        ),

        "reserved_gib": round(
            torch.cuda.memory_reserved(
                device_index
            ) / (1024 ** 3),
            3,
        ),

        "peak_allocated_gib": round(
            torch.cuda.max_memory_allocated(
                device_index
            ) / (1024 ** 3),
            3,
        ),

        "peak_reserved_gib": round(
            torch.cuda.max_memory_reserved(
                device_index
            ) / (1024 ** 3),
            3,
        ),
    }
    for device_index in range(
        torch.cuda.device_count()
    )
]


number_of_layers = int(
    model_config.num_hidden_layers
)

hidden_size = int(
    model_config.hidden_size
)

attention_heads = int(
    model_config.num_attention_heads
)

key_value_heads = int(
    getattr(
        model_config,
        "num_key_value_heads",
        attention_heads,
    )
)

head_dimension = (
    hidden_size
    // attention_heads
)

bytes_per_element = 2


layers_by_device = {}

for module_name, device in device_map.items():
    if module_name.startswith(
        "model.layers."
    ):
        device_name = str(device)

        layers_by_device[
            device_name
        ] = (
            layers_by_device.get(
                device_name,
                0,
            )
            + 1
        )


kv_bytes_per_layer_at_full_limit = (
    2
    * key_value_heads
    * head_dimension
    * configured_total_tokens
    * bytes_per_element
)


memory_feasibility = []

for memory_record in (
    memory_before_long_probe
):
    device_name = str(
        memory_record["index"]
    )

    layer_count = (
        layers_by_device.get(
            device_name,
            0,
        )
    )

    estimated_kv_bytes = (
        kv_bytes_per_layer_at_full_limit
        * layer_count
    )

    conservative_reserve_bytes = (
        estimated_kv_bytes * 2
    )

    memory_feasibility.append({
        "device": device_name,

        "assigned_layers": (
            layer_count
        ),

        "estimated_full_kv_cache_gib": round(
            estimated_kv_bytes
            / (1024 ** 3),
            3,
        ),

        "conservative_reserve_gib": round(
            conservative_reserve_bytes
            / (1024 ** 3),
            3,
        ),

        "free_before_probe_gib": round(
            memory_record[
                "free_estimate_bytes"
            ]
            / (1024 ** 3),
            3,
        ),

        "pass": (
            memory_record[
                "free_estimate_bytes"
            ]
            >= conservative_reserve_bytes
        ),
    })


long_prompt_close_to_limit = (
    long_input_token_count
    == MAX_RENDERED_INPUT_TOKENS
)

memory_estimate_pass = all(
    record["pass"]
    for record in memory_feasibility
)


long_context_status = (
    "PASS"
    if (
        long_prompt_close_to_limit
        and long_probe_generated_tokens == MAX_NEW_TOKENS
        and context_limit_pass
        and memory_estimate_pass
    )
    else "BLOCKED"
)


long_context_record = {
    "status": long_context_status,

    "prefill_chunk_size": (
        PREFILL_CHUNK_SIZE
    ),

    "synthetic_description_repetitions": (
        best_repetition_count
    ),

    "rendered_input_tokens": (
        long_input_token_count
    ),

    "configured_input_limit": (
        MAX_RENDERED_INPUT_TOKENS
    ),

    "configured_output_limit": (
        MAX_NEW_TOKENS
    ),

    "configured_total_limit": (
        configured_total_tokens
    ),

    "declared_context_window": (
        declared_context_window
    ),

    "generated_probe_tokens": (
        long_probe_generated_tokens
    ),

    "probe_duration_seconds": round(
        long_probe_duration_seconds,
        3,
    ),

    "prompt_sha256": sha256_bytes(
        best_long_render[
            "rendered_text"
        ].encode("utf-8")
    ),

    "model_architecture": {
        "layers": number_of_layers,
        "hidden_size": hidden_size,
        "attention_heads": (
            attention_heads
        ),
        "key_value_heads": (
            key_value_heads
        ),
        "head_dimension": (
            head_dimension
        ),
        "cache_dtype_bytes": (
            bytes_per_element
        ),
    },

    "layers_by_device": (
        layers_by_device
    ),

    "memory_feasibility": (
        memory_feasibility
    ),

    "observed_gpu_memory": (
        memory_after_long_probe
    ),

    "qualification_note": (
        "The frozen full input and continuation budgets were exercised "
        "using chunked prefill and forced continuation length on a synthetic prompt. "
        "The continuation is a capacity check only."
    ),
}


long_context_path = (
    PROJECT_ROOT
    / "reports"
    / "runtime"
    / "context_memory_qualification.json"
)

write_json(
    long_context_path,
    long_context_record,
)


print(
    json.dumps(
        long_context_record,
        indent=2,
    )
)


# Release probe-only tensors and cache.
del long_probe_output
del long_input_ids
del long_attention_mask

gc.collect()

for device_index in range(
    torch.cuda.device_count()
):
    torch.cuda.empty_cache()


if long_context_status != "PASS":
    raise RuntimeError(
        "Long-context qualification failed."
    )

print("\nCELL 17: PASS")
````

````python
def is_transient_file(
    path: Path,
) -> bool:
    return (
        "__pycache__" in path.parts
        or path.name == ".DS_Store"
        or path.suffix in {
            ".pyc",
            ".pyo",
        }
    )


manifest_input_roots = [
    PROJECT_ROOT / "config",
    PROJECT_ROOT / "reports" / "runtime",
    (
        PROJECT_ROOT
        / "data"
        / "development_examples"
    ),
]


artifact_paths = []

for input_root in manifest_input_roots:
    if not input_root.exists():
        continue

    if input_root.is_file():
        artifact_paths.append(
            input_root
        )

    else:
        artifact_paths.extend(
            path
            for path in input_root.rglob("*")
            if (
                path.is_file()
                and not is_transient_file(
                    path
                )
            )
        )


artifact_paths = sorted(
    set(artifact_paths),
    key=lambda path: (
        path.relative_to(
            PROJECT_ROOT
        ).as_posix()
    ),
)


artifact_records = [
    {
        "artifact_path": (
            path.relative_to(
                PROJECT_ROOT
            ).as_posix()
        ),
        "sha256": sha256_file(
            path
        ),
        "bytes": path.stat().st_size,
        "schema_version": "1.0",
        "producer_notebook": (
            "00_protocol_and_runtime"
        ),
    }
    for path in artifact_paths
]


configuration_paths = sorted(
    (
        path
        for path in (
            PROJECT_ROOT / "config"
        ).rglob("*")
        if path.is_file()
    ),
    key=lambda path: (
        path.relative_to(
            PROJECT_ROOT
        ).as_posix()
    ),
)


configuration_hash_payload = b"".join(
    (
        path.relative_to(
            PROJECT_ROOT
        ).as_posix().encode("utf-8")
        + b"\0"
        + path.read_bytes()
    )
    for path in configuration_paths
)

configuration_hash = sha256_bytes(
    configuration_hash_payload
)


stage_00_manifest = {
    "schema_version": "1.0",

    "study_id": STUDY_ID,
    "protocol_version": (
        PROTOCOL_VERSION
    ),

    "stage": "00",

    "producer_notebook": (
        "00_protocol_and_runtime"
    ),

    "created_utc": datetime.now(
        timezone.utc
    ).isoformat(),

    "stage_status": (
        stage_00_status
    ),

    "configuration_hash": (
        configuration_hash
    ),

    "upstream_manifests": [],

    "qualification": (
        stage_00_gate
    ),

    "artifact_count": len(
        artifact_records
    ),

    "artifacts": artifact_records,

    "notebook_source_location": (
        "Preserved by the corresponding "
        "Kaggle Notebook version."
    ),

    "model_snapshot_policy": (
        "Model weights are not copied into "
        "the Stage 00 ZIP. Their immutable "
        "revision and per-file hashes are "
        "recorded in model_snapshot_files.json."
    ),
}


stage_00_manifest_path = (
    PROJECT_ROOT
    / "registries"
    / "00_manifest.json"
)

write_json(
    stage_00_manifest_path,
    stage_00_manifest,
)


verification_problems = []

for artifact in artifact_records:
    artifact_path = (
        PROJECT_ROOT
        / artifact[
            "artifact_path"
        ]
    )

    if not artifact_path.is_file():
        verification_problems.append(
            "missing:"
            + artifact[
                "artifact_path"
            ]
        )
        continue

    if (
        artifact_path.stat().st_size
        != artifact["bytes"]
    ):
        verification_problems.append(
            "size_mismatch:"
            + artifact[
                "artifact_path"
            ]
        )

    if (
        sha256_file(artifact_path)
        != artifact["sha256"]
    ):
        verification_problems.append(
            "hash_mismatch:"
            + artifact[
                "artifact_path"
            ]
        )


cell_21_report = {
    "stage_status": (
        stage_00_status
    ),

    "manifest_path": str(
        stage_00_manifest_path
    ),

    "manifest_sha256": sha256_file(
        stage_00_manifest_path
    ),

    "artifact_count": len(
        artifact_records
    ),

    "artifact_bytes": sum(
        artifact["bytes"]
        for artifact
        in artifact_records
    ),

    "configuration_hash": (
        configuration_hash
    ),

    "verification_problems": (
        verification_problems
    ),

    "status": (
        "PASS"
        if not verification_problems
        else "BLOCKED"
    ),
}


print(
    json.dumps(
        cell_21_report,
        indent=2,
    )
)

if verification_problems:
    raise RuntimeError(
        "Manifest verification failed:\n"
        + "\n".join(
            verification_problems
        )
    )

print("\nCELL 21: PASS")
````

````python
import stat
import zipfile


archive_path = (
    DATASET_OUTPUT_ROOT
    / "stage_00_protocol_runtime_v1.zip"
)


archive_member_paths = sorted(
    set(
        artifact_paths
        + [stage_00_manifest_path]
    ),
    key=lambda path: (
        path.relative_to(
            PROJECT_ROOT
        ).as_posix()
    ),
)


descriptor, temporary_archive_name = (
    tempfile.mkstemp(
        dir=DATASET_OUTPUT_ROOT,
        prefix=".stage_00_archive.",
        suffix=".tmp",
    )
)

os.close(descriptor)

temporary_archive_path = Path(
    temporary_archive_name
)


try:
    with zipfile.ZipFile(
        temporary_archive_path,
        mode="w",
        compression=(
            zipfile.ZIP_DEFLATED
        ),
        compresslevel=9,
    ) as archive:
        for file_path in (
            archive_member_paths
        ):
            relative_path = (
                file_path
                .relative_to(
                    PROJECT_ROOT
                )
                .as_posix()
            )

            archive_name = (
                "prose_operation_experiment/"
                + relative_path
            )

            zip_information = (
                zipfile.ZipInfo(
                    filename=archive_name,
                    date_time=(
                        1980,
                        1,
                        1,
                        0,
                        0,
                        0,
                    ),
                )
            )

            zip_information.compress_type = (
                zipfile.ZIP_DEFLATED
            )

            zip_information.external_attr = (
                (
                    stat.S_IFREG
                    | 0o644
                )
                << 16
            )

            zip_information.create_system = 3

            archive.writestr(
                zip_information,
                file_path.read_bytes(),
                compress_type=(
                    zipfile.ZIP_DEFLATED
                ),
                compresslevel=9,
            )

    os.replace(
        temporary_archive_path,
        archive_path,
    )

except Exception:
    try:
        temporary_archive_path.unlink()
    except FileNotFoundError:
        pass

    raise


archive_sha256 = sha256_file(
    archive_path
)


external_manifest = {
    "schema_version": "1.0",

    "stage": "00",

    "producer_notebook": (
        "00_protocol_and_runtime"
    ),

    "status": stage_00_status,

    "archive": archive_path.name,

    "archive_sha256": (
        archive_sha256
    ),

    "archive_bytes": (
        archive_path.stat().st_size
    ),

    "internal_manifest_sha256": (
        sha256_file(
            stage_00_manifest_path
        )
    ),

    "model_id": MODEL_ID,

    "model_revision": (
        resolved_revision
    ),

    "corpus_source_url": (
        CORPUS_DATASET_URL
    ),

    "corpus_root": str(
        CORPUS_ROOT
    ),

    "upstream_datasets": [],
}


external_manifest_path = (
    DATASET_OUTPUT_ROOT
    / "stage_00_manifest.json"
)

checksum_path = (
    DATASET_OUTPUT_ROOT
    / "stage_00_checksums.sha256"
)

dataset_readme_path = (
    DATASET_OUTPUT_ROOT
    / "README.md"
)


write_json(
    external_manifest_path,
    external_manifest,
)

write_text(
    checksum_path,
    (
        f"{archive_sha256}  "
        f"{archive_path.name}\n"
    ),
)

write_text(
    dataset_readme_path,
    (
        "# Stage 00 — protocol and runtime\n\n"
        f"Status: **{stage_00_status}**\n\n"
        "This dataset contains the Stage 00 "
        "runtime and protocol archive for the "
        "prose-operation experiment.\n\n"
        f"Archive: `{archive_path.name}`\n\n"
        f"SHA-256: `{archive_sha256}`\n\n"
        "Read `stage_00_manifest.json` before "
        "using the archive. A NEEDS_REVIEW "
        "archive is diagnostic only; use the "
        "verified PASS archive from the recorded Kaggle "
        "run for downstream work.\n"
    ),
)


with zipfile.ZipFile(
    archive_path,
    mode="r",
) as archive:
    bad_member = archive.testzip()

    archive_names = (
        archive.namelist()
    )


expected_archive_names = [
    (
        "prose_operation_experiment/"
        + path.relative_to(
            PROJECT_ROOT
        ).as_posix()
    )
    for path in archive_member_paths
]


zip_verification_pass = (
    bad_member is None
    and archive_names
    == expected_archive_names
)


cell_22_report = {
    "stage_status": (
        stage_00_status
    ),

    "archive_path": str(
        archive_path
    ),

    "archive_sha256": (
        archive_sha256
    ),

    "archive_bytes": (
        archive_path.stat().st_size
    ),

    "archive_member_count": len(
        archive_names
    ),

    "zip_crc_failure": (
        bad_member
    ),

    "member_list_matches": (
        archive_names
        == expected_archive_names
    ),

    "external_manifest_path": str(
        external_manifest_path
    ),

    "checksum_path": str(
        checksum_path
    ),

    "readme_path": str(
        dataset_readme_path
    ),

    "status": (
        "PASS"
        if zip_verification_pass
        else "BLOCKED"
    ),
}


print(
    json.dumps(
        cell_22_report,
        indent=2,
    )
)

if not zip_verification_pass:
    raise RuntimeError(
        "Stage 00 ZIP verification failed."
    )

print("\nCELL 22: PASS")
````
